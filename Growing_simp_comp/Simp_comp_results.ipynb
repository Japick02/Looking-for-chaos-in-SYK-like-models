```python
import numpy as np
import matplotlib.pyplot as plt

import seaborn as sns

sns.set_theme(style="whitegrid")
plt.rcParams.update({
    "text.usetex": False,
    "font.family": "serif",
    "axes.labelsize": 12,
    "font.size": 12,
    "legend.fontsize": 10,
    "xtick.labelsize": 10,
    "ytick.labelsize": 10,
})
```

## Risultati con N=16

```python
N=16
```

```python
dati = np.load('simulation_data_N16_500x500.npz')

degree_var_list_1 = dati['variance']
max_bet_list_1 = dati['betweenness']
r_est_list_1 = dati['r']
sigma_est_list_1 = dati['sigma_r']
r_index_est_list_1 = dati['r_tilde']
sigma_index_est_list_1 = dati['sigma_r_tilde']
```

```python
dati = np.load('simulation2_data_N16_500x100.npz')

degree_var_list_2 = dati['variance']
max_bet_list_2 = dati['betweenness']
r_est_list_2 = dati['r']
sigma_est_list_2 = dati['sigma_r']
r_index_est_list_2 = dati['r_tilde']
sigma_index_est_list_2 = dati['sigma_r_tilde']
```

```python
degree_var_list = np.concatenate((degree_var_list_1, degree_var_list_2))
max_bet_list = np.concatenate((max_bet_list_1, max_bet_list_2))
r_est_list = np.concatenate((r_est_list_1, r_est_list_2))
sigma_est_list = np.concatenate((sigma_est_list_1, sigma_est_list_2))
r_index_est_list = np.concatenate((r_index_est_list_1, r_index_est_list_2))
sigma_index_est_list = np.concatenate((sigma_index_est_list_1, sigma_index_est_list_2))
N_com_list = dati['N_com']
modularity_list = dati['modularity']
```

```python
plt.figure(figsize=(9, 6), dpi=100)

plt.scatter(max_bet_list, r_index_est_list, 
            color='royalblue', 
            alpha=0.7, 
            edgecolors='black', 
            s=50, 
            label='Networks samples')

plt.axhline(y=0.386, color='crimson', linestyle='--', alpha=0.7, label='Poisson (0.386)')
plt.axhline(y=0.530, color='forestgreen', linestyle='--', alpha=0.7, label='GOE (0.531)')
plt.axhline(y=0.599, color='darkorange', linestyle='--', alpha=0.7, label='GUE (0.600)')

plt.xlabel('Max Link Betweenness Centrality', fontsize=12)
plt.ylabel(fr'$\langle \tilde r \rangle$', fontsize=12)
plt.title(fr'N={N}', fontsize=14, fontweight='bold')


plt.ylim(0.30, 0.65)


plt.grid(True, linestyle=':', alpha=0.6)


plt.legend(loc='lower right', fontsize=10)


plt.tight_layout()
plt.show()


plt.figure(figsize=(9, 6), dpi=100)

plt.scatter(degree_var_list, r_index_est_list, 
            color='royalblue', 
            alpha=0.7, 
            edgecolors='black', 
            s=50, 
            label='Networks samples')

plt.axhline(y=0.386, color='crimson', linestyle='--', alpha=0.7, label='Poisson (0.386)')
plt.axhline(y=0.530, color='forestgreen', linestyle='--', alpha=0.7, label='GOE (0.531)')
plt.axhline(y=0.599, color='darkorange', linestyle='--', alpha=0.7, label='GUE (0.600)')

plt.xlabel('Network Degree variance', fontsize=12)
plt.ylabel(fr'$\langle \tilde r \rangle$', fontsize=12)
plt.title(fr'N={N}', fontsize=14, fontweight='bold')

plt.ylim(0.30, 0.65)

plt.grid(True, linestyle=':', alpha=0.6)

plt.legend(loc='lower left', fontsize=10)

plt.tight_layout()
plt.show()

```

```python
plt.figure(figsize=(9, 6), dpi=100)


plt.scatter(modularity_list, r_index_est_list[500:], 
            color='royalblue', 
            alpha=0.7, 
            edgecolors='black', 
            s=50, 
            label='Networks samples')

plt.axhline(y=0.386, color='crimson', linestyle='--', alpha=0.7, label='Poisson (0.386)')
plt.axhline(y=0.530, color='forestgreen', linestyle='--', alpha=0.7, label='GOE (0.531)')
plt.axhline(y=0.599, color='darkorange', linestyle='--', alpha=0.7, label='GUE (0.600)')

plt.xlabel('Modularity', fontsize=12)
plt.ylabel(fr'$\langle \tilde r \rangle$', fontsize=12)
plt.title(fr'N={N}', fontsize=14, fontweight='bold')


plt.ylim(0.30, 0.65)

# Griglia di sfondo per leggere meglio i valori
plt.grid(True, linestyle=':', alpha=0.6)

# Posizionamento della legenda
plt.legend(loc='lower right', fontsize=10)

# 4. Mostra il grafico a schermo
plt.tight_layout()
plt.show()

```
